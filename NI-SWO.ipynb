```python
import numpy as np
import matplotlib.pyplot as plt
import math
import pandas as pd
```

```python
vals = [i / 100 for i in range(101)]  # 0.00, 0.01, ..., 1.00

h_pairs = [(h1, h2) for h1 in vals for h2 in vals]
```

```python
def W1(L1, L2, R1, c1, R2, c2, P, mu, h1, h2):
    return (L1*h1 + L2*h2)/(mu*(mu-L1*h1))
```

```python
def W2(L1, L2, R1, c1, R2, c2, P, mu, h1, h2):
    return (L1*h1 + L2*h2)/((mu - L1*h1 - L2*h2)*(mu-L1*h1))
```

```python
def cum_U1(L1, L2, R1, c1, R2, c2, P, mu, h1, h2):
    return L1*h1*(R1 - c1*W1(L1, L2, R1, c1, R2, c2, P, mu, h1, h2))
```

```python
def cum_U2(L1, L2, R1, c1, R2, c2, P, mu, h1, h2):
    return L2*h2*(R2 - c2*W2(L1, L2, R1, c1, R2, c2, P, mu, h1, h2))
```

```python
def max_util(L1, L2, R1, c1, R2, c2, P, mu, h_pairs):
    util_list = {}

    for h in h_pairs:
        h1, h2 = h

        U1 = cum_U1(L1, L2, R1, c1, R2, c2, P, mu, h1, h2)
        U2 = cum_U2(L1, L2, R1, c1, R2, c2, P, mu, h1, h2)
        U  = U1 + U2

        util_list[h] = (U, U1, U2)

    # h pair that yields the maximum social utility
    h_star = max(util_list, key=lambda h: util_list[h][0])
    h1, h2 = h_star[0], h_star[1]
    U_max, U1_max, U2_max  = util_list[h_star]

    return h1, h2, U1_max, U2_max



    
```

```python
max_util(0.125, 0.375, 10, 20, 3, 2.0, 0, 1, h_pairs)
```
